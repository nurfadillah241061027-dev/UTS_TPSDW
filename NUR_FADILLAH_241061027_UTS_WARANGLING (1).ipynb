# **install & import library**

```python
# Install library yang diperlukan
!pip install requests beautifulsoup4 pandas
# Import library
import requests
from bs4 import BeautifulSoup
import pandas as pd
import json
import time
```

# **Fungsi Scraping**

```python
base_url = "https://deepublishstore.com/product-category/ekonomi/page/{}/"

data = []

for page in range(1, 5):
    url = base_url.format(page)
    res = requests.get(url)
    soup = BeautifulSoup(res.text, "html.parser")

    books = soup.find_all("li", class_="product")

    for book in books:
        # Judul buku
        title = book.find("h2", class_="woocommerce-loop-product__title")
        title = title.text.strip() if title else ""

        # Harga buku
        price = book.find("span", class_="woocommerce-Price-amount amount")
        price = price.text.strip() if price else ""

        # Gambar buku
        img = book.find("img")
        img = img["src"] if img else ""

        data.append({
            "judul": title,
            "harga": price,
            "gambar": img
        })

df = pd.DataFrame(data)
df.head()
```

# **Simpan ke JSON (List_Buku.json)**

```python
df.to_json("List_Buku.json", orient="records", indent=4)
```

# **ETL (Ubah ke CSV)**

**Transform (bersihkan harga)**

```python
df["harga"] = df["harga"].str.replace("Rp", "")
df["harga"] = df["harga"].str.replace(".", "")
df["harga"] = df["harga"].str.strip()
```
